```python
import pandas as pd
import numpy as np
import joblib

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score


print("loading..")

# Load processed data
data = pd.read_csv("../data/processed_data.csv")


data = pd.get_dummies(data,columns=["Type", "Store", "IsHoliday"],drop_first=True)


# Features and target
X = data.drop(["Weekly_Sales", "Date"], axis=1)
y = data["Weekly_Sales"]

split_index = int(len(data) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]


```

```python
# Load trained models
l_model = joblib.load("../models/linear_regression.pkl")
best_rf = joblib.load("../models/best_random_forest.pkl")
best_xgb = joblib.load("../models/best_xgboost.pkl")


print("run successfully")


```

```python
# Predictions
l_pred = l_model.predict(X_test)
rf_pred = best_rf.predict(X_test)
xgb_pred = best_xgb.predict(X_test)


# Model Comparison
comparison = pd.DataFrame({
    "Model": [
        "Linear Regression",
        "Random Forest",
        "XGBoost"
    ],

    "MAE": [
        mean_absolute_error(y_test, l_pred),
        mean_absolute_error(y_test, rf_pred),
        mean_absolute_error(y_test, xgb_pred)
    ],

    "MSE": [
        mean_squared_error(y_test, l_pred),
        mean_squared_error(y_test, rf_pred),
        mean_squared_error(y_test, xgb_pred)
    ],

    "R2 Score": [
        r2_score(y_test, l_pred),
        r2_score(y_test, rf_pred),
        r2_score(y_test, xgb_pred)
    ]
})

print(comparison)
```
