```python

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import numpy as np
import pandas as pd
import joblib


data = pd.read_csv("../data/processed_data.csv")

data = data.sort_values("Date").reset_index(drop=True)

data = pd.get_dummies(data,columns=["Type", "Store", "IsHoliday"],drop_first=True)


X = data.drop(["Weekly_Sales", "Date"], axis=1)
y = data["Weekly_Sales"]


split_index = int(len(data) * 0.8)

X_train = X.iloc[:split_index]
X_test = X.iloc[split_index:]

y_train = y.iloc[:split_index]
y_test = y.iloc[split_index:]

l_model = LinearRegression()
l_model.fit(X_train, y_train)
l_pred = l_model.predict(X_test)


joblib.dump(l_model, "../models/linear_regression.pkl")

print("Linear Regression model saved!")


l_mae = mean_absolute_error(y_test, l_pred)

l_mse = mean_squared_error(y_test, l_pred)


l_r2 = r2_score(y_test, l_pred)

print(" MSE:", l_mse)
print(" MAE:", l_mae)
print(" R2 Score:", l_r2)
```
